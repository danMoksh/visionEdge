```python

from ultralytics import YOLO

model = YOLO(r"C:\Users\labca\FO's\model_paths\yolov8m.pt")  # or use your fine-tuned checkpoint

try:
    model.train(
        data=r"C:\Users\labca\Downloads\foreign object debris.v1i.yolov5pytorch\data.yaml",
        epochs=200,
        imgsz=640,
        batch=8,
        optimizer="AdamW",
        lr0=0.002,
        cos_lr=True,
        patience=20,
        device=0,
        amp=True,
        mosaic=1.0,
        hsv_h=0.015,
        hsv_s=0.7,
        hsv_v=0.4,
        flipud=0.0,
        fliplr=0.5,
        erasing=0.4,
        translate=0.1,       # NEW
        scale=0.5,           # NEW
        cls=0.3,             # NEW
        # obj=0.7,             # NEW
        # fl_gamma=2.0,        # NEW - focal loss
        iou=0.2,             # NEW - strict box matching
        workers=8,
        project="runs/detect",
        name="FOD_v1",
        save=True,
        plots=True,
        val=True,
        cache=False
    )
except Exception as e:
    print(f"[!] Crash caught: {e}")
    model.save("rescue_backup.pt")

    
```
